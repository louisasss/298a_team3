```python

import json

dependency_files = glob.glob(
    "./scripts/minimized-files/**/*dependencies.json",
    recursive=True
)

print("Dependency files count:", len(dependency_files))
print(dependency_files[:10])
```

## Check what is in a dependency file

```python
first_path = dependency_files[0]

with open(first_path, "r") as f:
    data = json.load(f)

print("File path for first dependency:", first_path)
print("Type:", type(data))
print(data)
```

```python
# For easier analysis of the data
print("Data keys: ", data.keys())
```

## Check the values in each feature

```python
print("Time:", data["time"])
print("Expected Run:", data["expectedRun"])
print("Expected:", data["expected"])
print("Dependent Test:", data["dependentTest"])
print("Polluters:", data["polluters"])
print("Hash:", data["hash"])
print("Flaky Class:", data["flakyClass"])
```

- OD means order-dependent flaky test.
- testOrder gives us the order of how we should be running the test.
- dependentTest is what we are investigating to see if it is flaky.

## Check the fields in expectedRun and Polluters

```python
print("Expected Run fields:", data["expectedRun"].keys())
print("Polluters fields:", data["polluters"][0].keys())
```

```python
print(data["polluters"][0]["cleanerData"].keys())
```

polluters give us more detail into what tests may change this behavior. (in this situation, it is testBindNested)

## Verify that all the files have the same values

```python
for path in dependency_files[:10]:
    with open(path, "r") as f:
        d = json.load(f)

    print(path)
    print(d.keys())
    print()
```

## Dependency file for all 6 json files

```python
rows = []

for path in dependency_files:
    with open(path, "r") as f:
        d = json.load(f)

    rows.append({
        "dependent_test": d["dependentTest"],
        "flaky_class": d["flakyClass"],
        "expected_result": d["expected"],
        "num_polluters": len(d["polluters"]),
        "num_tests": len(d["expectedRun"]["testOrder"]),
        "elapsed_seconds": d["time"]["elapsedSeconds"]
    })

summary_df = pd.DataFrame(rows)

summary_df
```
